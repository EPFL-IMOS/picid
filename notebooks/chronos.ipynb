```python
import pandas as pd

# Example time series data
data = pd.DataFrame({
    'timestamp': pd.date_range(start='2023-01-01', periods=1024, freq='D'),
    'value': range(1024)
})

# Ensure 'timestamp' is set as the DataFrame index
data.set_index('timestamp', inplace=True)

# Extract embeddings
embeddings = pipeline.embed(torch.tensor(data['value']))
```
