```python
import faiss
import numpy as np

# Step 1: Create some random high-dimensional vectors
d = 128  # Dimensionality
nb = 10000  # Database size
nq = 5  # Number of query vectors

np.random.seed(42)
database_vectors = np.random.random((nb, d)).astype('float32')
query_vectors = np.random.random((nq, d)).astype('float32')

# Step 2: Create FAISS index (L2 IndexFlat)
index = faiss.IndexFlatL2(d)  # L2 (Euclidean) distance index
print(f"Is trained? {index.is_trained}")

# Step 3: Add vectors to the index
index.add(database_vectors)
print(f"Total vectors in index: {index.ntotal}")

# Step 4: Perform a search for the k nearest neighbors
k = 10  # Number of nearest neighbors to retrieve
distances, indices = index.search(query_vectors, k)

# Step 5: Print results
print("\nNearest Neighbors (indices):")
print(indices)
print("\nDistances to Neighbors:")
print(distances)
```
