```python
import hydra
from pathlib import Path
from picid.transforms.base.transform_manager import ConfigTransformManager
from picid.data.datasources.base.multi_source_loader import MultiSourceLoader
from picid.data.datasources.base.phmd_loader import PHMDMultiSourceLoader
from picid.data.datasources.base.single_source_loader import SingleSourceLoader
from picid.data.preprocessing.preprocessor import PreProcessor

from awkward import to_numpy
import numpy as np
import matplotlib.pyplot as plt


%load_ext autoreload
%autoreload 2

data_name = "hsf15/prognostics/lstm.yaml"

task_modes = ["cooler","valve","pump","accumulator"]

colors = {
    "cooler": "blue",
    "valve": "orange",
    "pump": "green",
    "accumulator": "red"
}
collected_targets = {}
for task_mode in task_modes:
    with hydra.initialize(version_base="1.3", config_path="../configs", job_name="run"):
        # print('Current config file: ', config_file)
        cfg = hydra.compose(
            config_name="run.yaml",
            overrides=[f"experiment={data_name}",f"datasource.task_mode={task_mode}"],
            return_hydra_config=True,
        )

        dataset_source: SingleSourceLoader | MultiSourceLoader = hydra.utils.instantiate(
            cfg.datasource
        )
        # Create manager
        transforms_manager = ConfigTransformManager(transforms_config=cfg.transforms)
        single_source_types = (
            SingleSourceLoader,
            PHMDMultiSourceLoader,
        )
        if isinstance(dataset_source, single_source_types):
            preprocessor_mode = "per_unit"

        elif isinstance(dataset_source, MultiSourceLoader):
            preprocessor_mode = "cross_unit"

        else:
            raise ValueError("Invalid dataset source type")

        preprocessor = PreProcessor(preprocessor_mode, dataset_source, transforms_manager)
        

        # Get cache dirs and flags
        data_cache_path = (
            cfg.paths.cache_path
            if cfg.cache.use_cache_after_loading or cfg.cache.use_cache_after_transfroms
            else None
        )
        data_library_part_path = (
            Path(cfg.paths.root_dir) / "picid/data/datasources"
            if cfg.cache.use_cache_after_loading or cfg.cache.use_cache_after_transfroms
            else None
        )
        transform_library_part_path = (
            Path(cfg.paths.root_dir) / "picid/transforms"
            if cfg.cache.use_cache_after_loading or cfg.cache.use_cache_after_transfroms
            else None
        )
        cache_preprocessed_flag = cfg.cache.use_cache_after_transfroms
        use_file_lock = cfg.cache.get("use_preprocessing_file_lock", False)
        file_lock_path = cfg.cache.get(
            "preprocessing_file_lock_path", "/tmp/picid_preprocess.lock"
        )

        
        # Run the preprocessing pipeline
        preprocessor.pipeline(
            data_cache_path=data_cache_path,
            data_library_part_path=data_library_part_path,
            transform_library_part_path=transform_library_part_path,
            cache_preprocessed=cache_preprocessed_flag,
        )

        # TODO: return_splits_on_first_level is not clea..
        data_dict = preprocessor.get_processed_data_dict(
            return_splits_on_first_level=True
        )
        
        meta_data_dict = preprocessor.get_meta_data_dict()

        # Collect targets
        # THis dataset per se has one target per one cycle, so we can just extract last element from each unit and see the label disctribution
        target = to_numpy(data_dict['train']['target'][0])[:, -1]
        collected_targets[task_mode] = target

        plt.hist(target, bins=30)
        plt.title(f'Task mode: {task_mode} - Target distribution')
        plt.show()

        plt.plot(range(len(target)), target, color=colors[task_mode])
        plt.title(f'Task mode: {task_mode} - Cycles vs Target')

        plt.xlabel('Cycle')
        plt.ylabel('Target')  
        plt.show()

# plot the all targets together
plt.figure(figsize=(10,6))
for task_mode, target in collected_targets.items():
    plt.plot(range(len(target)), target, label=task_mode, color=colors[task_mode])
    # Assign color to every task mode
plt.title('All Task modes - Cycles vs Target')
plt.xlabel('Cycle')
plt.ylabel('Target')
plt.legend()
plt.show()  
```

```python



def print_features(cycle, cols_to_print="All", normalize=False):
    columns = [
        "ce", "cp", "eps1", "fs1", "fs2", "ps1", "ps2", "ps3", 
        "ps4", "ps5", "ps6", "se", "ts1", "ts2", "ts3", "ts4", "vs1"
    ]

    one_cycle_np = to_numpy(cycle)

    # 1. Determine exactly which columns we want to process
    if cols_to_print == "All":
        target_cols = columns
    else:
        # Ensure target_cols only contains valid names
        target_cols = [c for c in cols_to_print if c in columns]

    num_plots = len(target_cols)
    
    if num_plots == 0:
        print("No valid columns to print.")
        return

    # 2. Setup subplots (squeeze=False ensures axs is always a list/array, even for 1 plot)
    fig, axs = plt.subplots(num_plots, 1, figsize=(10, 2 * num_plots), squeeze=False)
    
    # Flatten axs so we can iterate over it easily regardless of dimensions
    axs = axs.flatten()

    # 3. Iterate through the TARGET columns, not the range of indices
    for i, col_name in enumerate(target_cols):
        
        # Find the index of this column name in the original full list
        original_data_index = columns.index(col_name)
        
        # Get the data using the correct index
        col_data = one_cycle_np[:, original_data_index]

        # Normalize
        if normalize:
            denom = np.max(col_data) - np.min(col_data)
            if denom == 0:
                col_data_norm = col_data # Avoid division by zero for constant signals
            else:
                col_data_norm = (col_data - np.min(col_data)) / denom
        else:
            col_data_norm = col_data

        # Plot on the ith subplot
        axs[i].plot(col_data_norm)
        axs[i].set_title(f'Column {col_name}')
        axs[i].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


# print 10 random PS1 cycles
for _ in range(10):
    random_index = np.random.randint(0, len(data_dict['train']['features'][0]))
    print(f"Random PS1 Cycle {random_index}")
    one_cycle = data_dict['train']['features'][0][random_index]
    print_features(one_cycle, cols_to_print=["ps1"], normalize=False)
```

```python
import numpy as np
def window_split(X, unit_id, signal_cols, target_col, subsignal_length=1000, signal_max_length=20000):
    """
    Splits time-series data into fixed-length windows for training machine learning models.
    
    This function processes time-series data, splitting it into smaller, fixed-length segments (windows).
    It ensures that each segment corresponds to a specific unit identified by the `unit_id` column
    and aligns the target variable for each segment to the last value in the window. 
    
    Parameters:
    -----------
    X : pandas.DataFrame
        The input dataset containing the time-series data.
        Must include columns specified by `unit_id`, `signal_cols`, and `target_col`.
    
    unit_id : str
        The name of the column in `X` used to identify individual units in the dataset.
    
    signal_cols : list of str
        A list of column names in `X` representing the signal data (features) for each unit.
    
    target_col : str
        The name of the column in `X` representing the target variable to predict.
    
    subsignal_length : int, optional, default=1000
        The length of each window (number of time-steps) to extract from the signal.
    
    signal_max_length : int, optional, default=20000
        The maximum number of time-steps to consider per unit.
        Excess data is truncated from the beginning to fit within this limit.
    
    Returns:
    --------
    X : numpy.ndarray
        A 3D array of shape `(num_windows, subsignal_length, num_features)`, where:
            - `num_windows` is the total number of extracted windows across all units.
            - `subsignal_length` is the fixed length of each window.
            - `num_features` is the number of signal features.
    
    Y : numpy.ndarray
        A 1D array of shape `(num_windows,)` containing the target variable values
        for each window, corresponding to the last time-step in the window.
    
    Example:
    --------
    >>> import pandas as pd
    >>> df = pd.DataFrame({
    ...     'unit_id': [1, 1, 1, 2, 2, 2],
    ...     'signal1': [0.1, 0.2, 0.3, 0.5, 0.6, 0.7],
    ...     'signal2': [0.4, 0.5, 0.6, 0.8, 0.9, 1.0],
    ...     'target': [10, 20, 30, 40, 50, 60]
    ... })
    >>> X, Y = window_split(df, 'unit_id', ['signal1', 'signal2'], 'target', subsignal_length=2, signal_max_length=4)
    >>> print(X.shape)
    (2, 2, 2)
    >>> print(Y)
    [20. 50.]
    """
        
    units_array = X[unit_id].values
    signal_array = X[signal_cols].values
    targets = X[target_col].values
    N = signal_max_length // subsignal_length

    num_units = X[unit_id].drop_duplicates().shape[0]
    unique_units = X[unit_id].drop_duplicates().values
    X = np.zeros((N * num_units, subsignal_length, len(signal_cols)))
    Y = np.zeros((N * num_units,))
    NN = 0
    for i, unit in enumerate(unique_units):
        mask = np.all(units_array == unit, axis=1)
        signal = signal_array[mask]
        sml = (signal.shape[0] // subsignal_length) * subsignal_length
        n = min(sml // subsignal_length, N)
        sml = subsignal_length * n
        signal = signal[-sml:]
        signal = signal.reshape((n, subsignal_length, len(signal_cols)))
        signal_target = targets[-sml:]
        signal_target = signal_target.reshape((n, subsignal_length))
        X[NN:NN + n] = signal
        Y[NN:NN + n] = signal_target[:, -1]
        NN += n

    X = X[:NN]

    return X, Y
```

```python
def convert_outer_list_to_inner(data_list):
    if not data_list:
        return {}

    # Ensure all dictionaries have the same keys
    keys = data_list[0].keys()
    for d in data_list:
        if d.keys() != keys:
            raise ValueError("Not all dicts have the same keys!")

    # Concatenate values for each key
    stacked = {}
    for key in keys:
        stacked[key] = [d[key] for d in data_list]
        
    # Check that all keys in the stacked have the same length
    lengths = [len(stacked[key]) for key in stacked]
    if len(set(lengths)) != 1:
        raise ValueError("All keys in the stacked dict must have the same length.")
    
    return stacked  
```

```python
from collections import defaultdict
import numpy as np

for dataset in ['UNIBO21', 'NB1', 'NB14']:
    # Load dataset and metadata
    ds = datasets.Dataset(dataset, cache_dir='data')
    task = ds['rul']
    folds = task[0]  # fold_num = 0 by default

    unit_id = ds['rul'].meta['identifier']
    assert len(unit_id) == 1, "There should be only one identifier"
    unit_col = unit_id[0]
    features_col = ds['rul'].meta['features']
    target_col = 'rul'

    # Helper function to process a single unit
    def process_unit(df_unit, unit_name):
        df_features = df_unit[features_col].copy().reset_index(drop=True)
        df_target = df_unit[target_col].copy().reset_index(drop=True)

        if df_unit.isna().any().any():
            print(f"NaN values detected in unit {unit_name}. Details:\n{df_unit.isna()}")
            df_features = df_features.dropna()  # Optionally handle df_target too

        return {
            "features": df_features,
            "timestamps": np.array(df_features.index),
            "target": df_target,
            "columns": df_features.columns,
            "target_name": target_col,
            "metadata": {
                'unit_id': unit_name,
                'unit_length': len(df_features)
            }
        }

    # Process each split
    data = defaultdict(list)
    for split in ['train', 'val', 'test']:
        df = folds[split]
        for unit_name in np.unique(df[unit_col]):
            df_unit = df[df[unit_col] == unit_name]
            data[split].append(process_unit(df_unit, unit_name))

    # Convert to desired format
    train = convert_outer_list_to_inner(data['train'])
    val = convert_outer_list_to_inner(data['val'])
    test = convert_outer_list_to_inner(data['test'])

    out_dict = {}
    for key in train.keys():
        out_dict[key] = {
            "train": train[key],
            "val": val[key],
            "test": test[key],
            }
    print("Number of train units:", len(out_dict['features']['train']))
    print("Number of val units:", len(out_dict['features']['val']))
    print("Number of test units:", len(out_dict['features']['test']))
```
