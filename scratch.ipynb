```python
from numpy import ndarray
import numpy as np

def swap_first_second_level(data: dict[str, ndarray | list[ndarray]]) -> dict[str, ndarray | list[ndarray]]:
    """
    Swap the first and second level of a nested dictionary.
    """
    if not isinstance(data, dict):
        raise ValueError("Input must be a dictionary.")

    swapped = {}
    for key, value in data.items():
        if isinstance(value, dict):
            # If value is a dict, check if its values are also dicts (deeper than 2 levels)
            for sub_key, sub_value in value.items():
                if isinstance(sub_value, dict):
                    raise ValueError("Error while swapping dictionary. Dictionary is deeper than 2 levels.")
                if swapped.get(sub_key) is None:
                    swapped[sub_key] = {}
                swapped[sub_key][key] = sub_value
        else:
            raise ValueError("Error while swapping dictionary. First level value must be a dict.")

    return swapped




dictionary = {
    "key1": {
        "train": np.array([1, 2, 3]),
        "test": np.array([4, 5, 6])
    },
    "key2": {
        "train": np.array([7, 8, 9]),
        "test": np.array([10, 11, 12])
    }
}

swapped_dict = swap_first_second_level(dictionary)

print(swapped_dict)
```
